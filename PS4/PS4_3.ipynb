```python
def single_linear_regression(X,Y):
  n = len(X)

  sum_x = sum(X)
  sum_y = sum(Y)

  sum_xy = sum(X[i] * Y[i] for i in range(n))
  sum_x2 = sum(X[i]**2 for i in range(n))

  b1 = ((n*sum_xy) - (sum_x*sum_y))/((n*sum_xy)- (sum_x2**2))
  b0 = (sum_y - b1*sum_x)/n

  return b0,b1
```

```python
def predict_single(b0,b1,x):
  return b0+b1*x
```

```python
X = [1, 2, 3, 4, 5]
Y = [2, 4, 5, 4, 5]

b0, b1 = single_linear_regression(X, Y)

print("Learned Parameters:")
print("Intercept (b0):", b0)
print("Slope (b1):", b1)

# Test Data
test_x = 6
prediction = predict_single(b0, b1, test_x)
```

Multi regression

```python
import numpy as np

def multiple_linear_regression(X,Y):
  for i in range(len(X)):
    X[i] = [1] + X[i]

  Y = [[y] for y in Y]

  XT = np.transpose(X)
  XTX = np.dot(XT,X)
  XTX_inv = np.linalg.inv(XTX)
  XTY = np.dot(XT,Y)

  beta = np.dot(XTX_inv,XTY)

  return beta
```

```python
def predict_multiple(beta,x):
  x = [1] + x
  result = 0

  for i in range(len(beta)):
    result += beta[i][0]*x[i]
  return result
```

```python
X = [
    [1,2],
    [2,1],
    [3,4],
    [4,3],
    [5,5]
]
Y = [5,6,9,10,13]

beta = multiple_linear_regression(X,Y)

test_input = [6, 7]
prediction = predict_multiple(beta, test_input)

print("Prediction for", test_input, "is:", prediction)
```
