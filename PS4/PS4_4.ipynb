```python
import pandas as pd
import numpy as np
import math
from sklearn.model_selection import train_test_split
```

```python
# 1. Prep Data (Keep the original structure)
df = pd.read_csv('/content/drive/MyDrive/College/Sem 8/DM/CA2/play_tennis.csv')
df = pd.get_dummies(df, columns=['Outlook', 'Wind', 'PlayTennis'], dtype=int)
```

```python
y = df['PlayTennis_Yes']
X = df.drop(['PlayTennis_Yes','PlayTennis_No'],axis=1)
```

```python
# Scaling
for col in ["Temp_Value","Humidity_Value"]:
  X[col] = (X[col]-X[col].mean()) / X[col].std()

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=0)
```

```python
def sigmoid(z):
  return 1/(1+ np.exp(-z))
```

```python
def train_logistic_regression(x,y,learning_rate=0.1,epochs=10):
  n_samples,n_features = x.shape

  weights = [0.0] * (n_features + 1)

  for epoch in range(epochs):
    for i in range(n_samples):
      # Calculate Dot Product: z = b0 + b1*x1 + b2*x2...
      z = weights[0]
      for j in range(n_features):
        z += weights[j+1]*x.iloc[i,j]
      prediction = sigmoid(z)
      error = y.iloc[i] - prediction

      weights[0] += learning_rate * error
      for j in range(n_features):
        weights[j+1] += learning_rate*error*x.iloc[i,j]
    return weights

```

```python
def predict(x,weights):
  results = []
  n_features = x.shape[1]

  for i in range(len(x)):
    z = weights[0]

    for j in range(n_features):
      z += weights[j+1]*x.iloc[i,j]
    prob = sigmoid(z)
    results.append(1 if prob >= 0.5 else 0)
  return results
```

```python
# 3. Execution
trained_weights = train_logistic_regression(X_train, y_train)
y_pred = predict(X_test, trained_weights)

from sklearn.metrics import classification_report

print(classification_report(y_test,y_pred))
```
