```python
import numpy as np
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/College/Sem 8/DM/CA2/play_tennis.csv')
```

```python
for col in df.columns:
  if df[col].dtype == "O":
    unique_values = df[col].unique().tolist()
    df[col] = df[col].apply(lambda x : unique_values.index(x))
```

```python
X = df
for col in X.columns:
  X[col] = (X[col]-X[col].mean())/X[col].std()
X = X.values
```

```python
def kmeans(X,k):
  n_samples,n_features = X.shape
  indices = np.random.choice(len(X),k,replace=False)
  centroids = X[indices]

  for _ in range(100):
    clusters = []
    for x in X:
      distances = [np.sqrt(sum((x-c)**2)) for c in centroids]
      clusters_idx = np.argmin(distances)
      clusters.append(clusters_idx)

    new_centroids = []
    for i in range(k):
      points = X[clusters==i]
      if points.size > 0:
        new_centroids.append(points.mean(axis=0))
      else:
        new_centroids.append(centroids[i])

    new_centroids_arr = np.array(new_centroids)
    if np.allclose(new_centroids_arr, centroids):
      break

    centroids = new_centroids_arr

  return clusters, centroids.tolist()
```

```python
# 2. Execution
k = 3
labels, final_centroids = kmeans(X, k)

# Convert np.int64 labels to standard Python int
int_labels = [int(label) for label in labels]

print(f"Cluster Assignments: {int_labels}")
print(f"Final Centroids:\n {final_centroids}")
```
