```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
np.random.seed(42)

data = {
    'Gender': np.random.choice(['Male', 'Female'], 100),
    'Preference': np.random.choice(['Product A', 'Product B', 'Product C'], 100),
    'Age': np.random.randint(18, 70, 100),
    'Spend_Amount': np.random.uniform(50, 500, 100)
}
df = pd.DataFrame(data)

print("--- Original Dataset (First 5 Rows) ---")
print(df.head())
print()
```

```python
df_minmax = df.copy()

for col in ['Age', 'Spend_Amount']:
    min_val = df[col].min()
    max_val = df[col].max()
    df_minmax[col] = (df[col] - min_val) / (max_val - min_val)

print("--- Min-Max Normalized Data (First 5 Rows) ---")
print(df_minmax[['Age', 'Spend_Amount']].head())
print()
```

```python
df_zscore = df.copy()

for col in ['Age', 'Spend_Amount']:
    mean = df[col].mean()
    std = df[col].std()
    df_zscore[col] = (df[col] - mean) / std

print("--- Z-Score Normalized Data (First 5 Rows) ---")
print(df_zscore[['Age', 'Spend_Amount']].head())
print()
```
