```python
import pandas as pd
import numpy as np

# 1. Dataset Setup
data = pd.DataFrame({
    'Age': [19, 23, 34, 65, 78, 89],
    'Class': ['Yes', 'Yes', 'No', 'No', 'Yes', 'Yes']
})

def calculate_entropy(y):
    # P(class) = count / total
    probs = y.value_counts() / len(y)
    # Entropy = - sum(p * log2(p))
    return -np.sum(probs * np.log2(probs + 1e-9))

def find_best_numerical_split(df, attr, target):
    # Entropy of the entire dataset (D)
    parent_entropy = calculate_entropy(df[target])
    print(f"Entropy of D: {parent_entropy:.4f}\n")

    # Sort unique values and find midpoints (Split Points)
    vals = sorted(df[attr].unique())
    split_points = [(vals[i] + vals[i+1]) / 2 for i in range(len(vals) - 1)]

    best_gain = -1
    best_point = None

    print(f"{'Split Point':<15} | {'Weighted Entropy':<18} | {'Info Gain':<10}")
    print("-" * 50)

    for point in split_points:
        # Split data into two subsets
        d1 = df[df[attr] <= point]
        d2 = df[df[attr] > point]

        # Weighted Entropy = (n1/n)*Ent(D1) + (n2/n)*Ent(D2)
        n = len(df)
        w_entropy = (len(d1)/n) * calculate_entropy(d1[target]) + \
                    (len(d2)/n) * calculate_entropy(d2[target])

        # Information Gain = Entropy(D) - Weighted Entropy
        gain = parent_entropy - w_entropy

        print(f"{point:<15} | {w_entropy:<18.4f} | {gain:<10.4f}")

        if gain > best_gain:
            best_gain = gain
            best_point = point

    return best_point, best_gain

# 2. Execution
best_p, best_g = find_best_numerical_split(data, 'Age', 'Class')

print("-" * 50)
print(f"Best Split Point: {best_p}")
print(f"Max Information Gain: {best_g:.4f}")
```
