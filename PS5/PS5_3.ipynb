```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/drive/MyDrive/College/Sem 8/DM/CA2/play_tennis.csv')
```

```python
for col in df.columns:
  if df[col].dtype == 'O':
    unique_vals = df[col].unique().tolist()
    df[col] = df[col].apply(lambda x : unique_vals.index(x))
```

```python
target = df.columns[-1]
X = df.drop(target,axis = 1)
y = df[target].values
```

```python
for col in X.columns:
  X[col] = (X[col] - X[col].mean())/X[col].std()
X = X.values
```

```python
def predict_knn(X_train,y_train,X_test,k,task = "classification"):
  preds = []
  for x_test in X_test:
    distances = []
    for x_tr in X_train:
      d = np.sqrt(np.sum((x_test-x_tr)**2))
      distances.append(d)
    k_indices = np.argsort(distances)[:k]
    k_neighbours = [y_train[i] for i in k_indices]

    if task == 'classification':
      counts = {}
      for val in k_neighbours:
        if val not in counts.keys():
          counts[val]=1
        else:
          counts[val]+=1
      result = max(counts,key=counts.get)
      preds.append(result)
    else:
      preds.append(sum(k_neighbours)/k)

  return preds
```

```python
# 4. Usage
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

y_pred = predict_knn(X_train, y_train, X_test, k=3)

# Accuracy Calculation
acc = sum(1 for p, a in zip(y_pred, y_test) if p == a) / len(y_test)
print(f"Accuracy: {acc * 100:.2f}%")
```
