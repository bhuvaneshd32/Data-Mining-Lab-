```python
import numpy as np

distance_matrix = [
    [0, 2, 6, 10],
    [2, 0, 5, 9],
    [6, 5, 0, 4],
    [10, 9, 4, 0]
]
```

```python
def agglomerative_clustering(dist_mat,linkage,k=2):
  n = len(dist_mat)
  # Start: Every point is its own cluster [[0], [1], [2], [3]]
  clusters = [[i] for i in range(n)]

  while( len(clusters)> k):
    min_dist = float('inf')
    merge_idx = (0,1)
    for i in range(len(clusters)):
      for j in range(i+1,len(clusters)):
        pair_dists = [dist_mat[p][q] for p in clusters[i] for q in clusters[j]]
        dist = min(pair_dists) if linkage=="single" else max(pair_dists)

        if dist<min_dist:
          merge_idx = (i,j)
          min_dist = dist
    idx,idj = merge_idx
    new_cluster = clusters[idx]+clusters[idj]
    clusters.pop(idj)
    clusters.pop(idx)
    clusters.append(new_cluster)
  return clusters


```

```python
# Test
print("Single Link:", agglomerative_clustering(distance_matrix, "single"))
print("Complete Link:", agglomerative_clustering(distance_matrix, "complete"))
```
